## Bloque 2 — Ejecutar: Crear audiencia + compartir

```python
-- Crear nueva audiencia segmentada
CREATE OR REPLACE SECURE VIEW CREDIBANCO_HOL.MONETIZACION.AUD_TURISMO_<TU_USUARIO> AS
SELECT REGION, MCC, SUM(MONTO_TOTAL) AS MONTO, SUM(NUM_TRANSACCIONES) AS TXS
FROM CREDIBANCO_HOL.TURISMO.AGREGADOS_TURISMO
GROUP BY 1, 2
HAVING SUM(NUM_TRANSACCIONES) > 100;
```

```python
-- Agregar al share existente
GRANT SELECT ON VIEW CREDIBANCO_HOL.MONETIZACION.AUD_TURISMO_<TU_USUARIO>
  TO SHARE SH_AUDIENCIA_ALTO_VOLUMEN_USER;
```
